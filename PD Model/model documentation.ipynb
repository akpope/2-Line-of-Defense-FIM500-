#### *Brief Background* 

For the benefit of a banking insitution, it is important mitigate the potential credit loss from borrowers. Probability of default (PD) refers to the probability that a borrower will default on their loan within a certain time period, leading to a banking insitution suffering from credit loss. A credit risk model calculating the probability of default for current borrowers will decrease the financial risk within the banking organization, as precautions or specific measures can be taken once the probability of default is too high

#### *Problem Statement and Business Context*

The aim of this project is to create a PD model using logistic regression based on various features for one target variable. Loans, credit lines, and other various products within a bank are the ways banks generate a good amount of their capital/assets, however these products also carry risk associated with the quality of the borrowers. This is why it's important to assess the probability of default for borrowers who are already within the establishment of the organization, just because they've been approved for a loan doesn't mean their credit/other variables can't change throughout the lifetime of their credit lines and/or loans.

#### *Logistic Regression Model Details*

For logistic regression the following assumptions are made; there are only two possible outcomes (0 and 1), every single observation (borrower account in this case) is independent, and there exsists no multicollineatiry. Since logistic regression returns a probability, it is the best option to utilize as a model to predict probability of default. The target variable of this model is `SeriousDlqin2yrs`, while the features of this model are;
+ `RevolvingUtilizationOfUnsecuredLines` 
+ `NumberOfTime30-59DaysPastDueNotWorse`
+ `DebtRatio`
+ `MonthlyIncome`
+ `NumberOfTimes90DaysLate`
+ `NumberOfTime60-89DaysPastDueNotWorse`
<br>

Justification for the selection of the features are as follows;
+ Revolving Utilization shows how much a current borrower is using of their unsecured line/credit, meaning if they have a higher value of this ratio they are using more of the credit. A high utilization ratio can indicate that a borrower is relying heavily on the credit line and potentially have low financial stability/flexbility. From this we can infer that a borrower may default if they have a higher utilization ratio.
+ For all of the number of times days past due/days last this shows how late a borrower was on their obligated payment schedule. If a borrower has accumulated a lot of these delinquencies, it can be inferred that they do not have the ability to pay back their loans.
+ Debt ratio shows how much debt to income a current borrower has. A higher debt ratio indicates that a borrower has a lot of debt to take on, meaning they might not have the funds to cover the payments for a credit line/loan
+ Monthly income is a good measure of seeing if a borrower can repay/stick to their payment plan for credit lines/loans
<br>

The logistic model I am using is the LogisticRegression() function from the scikit-learn package on Python ([scikit-learn Logistic Regression](https://scikit-learn.org/0.21/modules/linear_model.html#logistic-regression)). #### *Pre-Validation Testing*

After finishing up the baseline model, it's important to do some pre-validation testing. For this, I used precision-recall curve, confusion matrix, and ROC-AUC analysis. 

To label everything as false negatives, true positives, false positives, and true negatives, I have to convert all the probabilities into a 1 (default within 2 years) or 0 (no default within 2 years) (similar to the target variable). From here, using the precision-recall curve, I can determine the proper cut-off/threshold to determine if there was a default within 2 years. The precision-recall curve that I got, shows that there's some issues within the model that I have created. The trade-off is quite low as the recall rate increases. 

![Precision-recall curve](precision_recall_curve.png)

Starting with the confusion matrix, to label everything as false negatives, true positives, false postiives, and true negatives, I have to convert all the probabilities into a 1 or 0 (similar to the target variable). The threshold to determine if there was a default within 2 years (1) was if the probability of default is greater than 25% (since from the precision-recall graph, this has the best trade off between the two). From the confusion matrix based on this threshold, I got results in the form of the recall, precision, and accuracy rates.

| Rate Type | Value (%) |
|---|---:|
| Recall | 34.68 |
| Precision | 31.32 |
| Accuracy | 90.55 |

The recall rate represents all of the positives that were classified correstly as true positives - from this low rate, it shows that there are a lot of false negatives, which is not good. That means the model is classifying people who should have defaulted/were shown to default as negative. This could result in credit loss within our banking organization if this model was implemented. As for the precision, it's representing the percent of true positives that were classified correctly compared to all positives (true and false positives). This rate is also pretty low, indicating that there are a lot of false positives. So people who were meant to not default, were predicted to default in this model. This incorrect classification isn't as damaging as a false negative, but it still impacts the time spent investigating an incorrect default. Finally, the accuracy score is pretty high, but since only 6% of the borrowers defaulted from the original data, my model detects a lot of the true negative cases. 

Next up was the ROC-AUC analysis. With the ROC curve, this shows the model performance across any threshold selection for determining the default within 2 years. Meanwhile the AUC, is the area under the ROC curve which represents the probability that the model correctly assigns the defaulters. The ROC curve along with AUC value is below,

![ROC-AUC curve](rocauc_curve.png)

Note that FPR is the false positive rate, while TPR is the true positive rate. With the AUC value being ~81%. The model has an ~81% chance of correctly ranking a randomly selected defaulter as having a higher probability of default than a randomly selected non-defaulter. So this model does a relatively decent job at distinguishing defaulters from non-defaulters based on the predicted default probabilities.
